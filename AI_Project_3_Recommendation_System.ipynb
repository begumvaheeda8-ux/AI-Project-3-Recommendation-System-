```python
# Create a simple movie dataset

data = {
    "Movie": [
        "Inception",
        "Interstellar",
        "The Dark Knight",
        "Titanic",
        "The Notebook",
        "Avengers",
        "Jurassic Park",
        "The Conjuring"
    ],
    "Genre": [
        "Sci-Fi",
        "Sci-Fi",
        "Action",
        "Romance",
        "Romance",
        "Action",
        "Adventure",
        "Horror"
    ],
    "Type": [
        "Mind-Bending",
        "Space",
        "Superhero",
        "Drama",
        "Drama",
        "Superhero",
        "Dinosaur",
        "Supernatural"
    ]
}

df = pd.DataFrame(data)

df
```

```python
# User preferences

user_genre = "Sci-Fi"
user_type = "Space"

print("User Genre Preference:", user_genre)
print("User Type Preference:", user_type)
```

```python
# Calculate similarity score

def calculate_similarity(row):
    score = 0

    if row["Genre"] == user_genre:
        score += 1

    if row["Type"] == user_type:
        score += 1

    return score


df["Similarity Score"] = df.apply(calculate_similarity, axis=1)

df
```

```python
# Sort movies by similarity score

recommendations = df.sort_values(
    by="Similarity Score",
    ascending=False
)

recommendations
```

```python
# Select top 3 recommendations
top_recommendations = recommendations.head(3)

print("Top 3 Movie Recommendations:")
print(top_recommendations[["Movie", "Genre", "Type", "Similarity Score"]])
```

```python
def recommend_movies(genre, movie_type):

    temp_df = df.copy()

    def similarity(row):
        score = 0

        if row["Genre"] == genre:
            score += 1

        if row["Type"] == movie_type:
            score += 1

        return score

    temp_df["Similarity Score"] = temp_df.apply(
        similarity,
        axis=1
    )

    result = temp_df.sort_values(
        by="Similarity Score",
        ascending=False
    )

    return result[[
        "Movie",
        "Genre",
        "Type",
        "Similarity Score"
    ]].head(3)
```

```python
# Test the recommendation system

recommend_movies("Sci-Fi", "Space")
```

```python
# Test with another user preference

recommend_movies("Romance", "Drama")
```
